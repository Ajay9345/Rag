```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader("/content/COMPLETE SPECIFICATION (1).pdf")
pdf = loader.load()
```

```python
pdf[0]
```

```python
len(pdf)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
            chunk_size=1024,
            chunk_overlap=200
        )
```

```python
chunked_pdf = splitter.split_documents(pdf)
```

```python
len(chunked_pdf)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
multilingual_embeddings = HuggingFaceEmbeddings(model_name='intfloat/multilingual-e5-large')
```

```python
multilingual_embeddings.embed_query(chunked_pdf[0].page_content)
```

```python
len(multilingual_embeddings.embed_query(chunked_pdf[0].page_content) )
```

```python
!pip install chromadb --quiet
```

```python
from langchain_community.vectorstores import Chroma
```

```python
vectordb = Chroma.from_documents(
            documents=chunked_pdf,
            embedding=multilingual_embeddings,
            persist_directory="/db/chroma/"
        )
```

```python
question = "summarize all the 10 claims"
```

```python
vectordb.similarity_search(question, k=3)
```

```python
vectordb.similarity_search_with_score(question, k=3)
```

```python
vectordb.max_marginal_relevance_search(question, k=3, fetch_k=10)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
system_prompt = (
        "You are an assistant for question-answering tasks. Use the following pieces of retrieved context to answer the question",
        "If you don't know the answer, just say that you don't know. Use three sentences maximum and keep the answer concise",
        "Answer all questions to the best of your ability",
        )
```

```python
retrieved_docs = vectordb.similarity_search(question, k=5)

context = " ".join([doc.page_content for doc in retrieved_docs])

system_prompt_as_string = " ".join(system_prompt)

messages = [
    SystemMessage(content=system_prompt_as_string),
    HumanMessage(
        content=f"""
Use the following context to answer the question.
If the answer is not in the context, say you don't know.

Context:
{context}

Question:
{question}
"""
    )
]

result = llm.invoke(messages)
result.content
```
